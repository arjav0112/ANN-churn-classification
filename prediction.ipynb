```python
import tensorflow as tf
import pandas as pd
from tensorflow.keras.models import load_model
import pickle
```

```python
model = load_model('model.h5')

with open('onehot_encode_geo.pkl','rb') as file:
    label_encode_geo = pickle.load(file)

with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
input_data = {
    'CreditScore' : 600,
    "Geography" : 'France',
    'Gender' : 'Male',
    'Age' : 40,
    'Tenure' : 3,
    'Balance' : 60000,
    'NumOfProducts' : 2,
    'HasCrCard' : 1,
    'IsActiveMember' : 1,
    'EstimatedSalary' : 50000
}
```

```python
geo_encoder = label_encode_geo.transform([[input_data['Geography']]]).toarray()
geo_encoder_df = pd.DataFrame(geo_encoder,columns=label_encode_geo.get_feature_names_out(['Geography']))
geo_encoder_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
import numpy as np

input_df['Gender'] = label_encoder_gender.fit_transform(input_df['Gender'])
input_df
```

```python
input_df = pd.concat([input_df.drop(['Geography'],axis=1), geo_encoder_df], axis=1)
input_df
```

```python
input_scale = scaler.transform(input_df)
input_scale
```

```python
predicton = model.predict(input_scale)
predicton
```

```python
predicton_prob = predicton[0][0]
predicton_prob
```

```python
if predicton_prob > 0.5:
    print("Customer will leave the bank")
else:   
    print("Customer will not leave the bank")
```
